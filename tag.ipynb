```python
import numpy as np
import pandas as pd

df = pd.read_excel("實時同步翻譯文本.xlsx")
df.replace(np.nan, None, inplace=True)  # replace np.nan with None
df
```

```python
import re

import pandas as pd

# 读取Excel文件
df_glossary = pd.read_excel("glossary_all.xlsx")

# 只保留需要的两列
df_glossary = df_glossary[["zh-CN", "ko"]]

# 删除任何包含NaN的行
df_glossary = df_glossary.dropna(how='any')

# 按照中文列去重
df_glossary = df_glossary.drop_duplicates(subset="zh-CN", keep=False)

# 确保'zh-CN'和'ko'列中的所有元素都是字符串
df_glossary['zh-CN'] = df_glossary['zh-CN'].astype(str)
df_glossary['ko'] = df_glossary['ko'].astype(str)

# 删除包含数字的行
mask = df_glossary.applymap(lambda x: bool(re.search(r'\d', x)))
df_glossary = df_glossary[~mask.any(axis=1)]

#按照长度降序排布
df_glossary["zh-CN_length"] = df_glossary["zh-CN"].str.len()
df_glossary = df_glossary.sort_values("zh-CN_length", ascending=False)
df_glossary = df_glossary.drop('zh-CN_length', axis=1)
df_glossary
```

```python
df_glossary.set_index("zh-CN", inplace=True)
glossary_dict = df_glossary["ko"].to_dict()
glossary_dict
```

```python
text = "[color=#F7C358][size=40]离线挂机特权[/color][/size]	[color=#F7C358][size=40]오프라인 자동전투 패스[/color][/size]	j界面文本.xlsx"
suffix_tags = list(set(re.findall("\[\/.*?\]", text)))  # e.g. ['[/color]', '[/size]']
suffix_tags
```

```python
suffix_tag_types = list(set(re.findall("\[\/(.*?)\]", text)))  # e.g.['size', 'color']
suffix_tag_types
```

```python
regex = "|".join([r"\[{0}=.*?\]".format(tag_type) for tag_type in suffix_tag_types])  # e.g.'\[size=.*?\]|\[color=.*?\]'
regex
```

```python
prefix_tags = list(set(re.findall(regex, text)))  # e.g. ['[color=#F7C358]', '[size=40]']
prefix_tags
```

```python
tags_all = suffix_tags + prefix_tags  # e.g. ['[/color]', '[/size]', '[color=#F7C358]', '[size=40]']
tags_all
```

```python
from tqdm.notebook import tqdm

for index, row in tqdm(df.iterrows(), total=df.shape[0]):
    for column in df.columns:
        if row[column] is not None:
            text = str(row[column])
            # HTML标签预处理
            suffix_tags = list(
                set(re.findall("|".join(["\[\/.*?\]", "\<\/.*?\>"]), text)))  # e.g. ['[/color]', '[/size]']
            suffix_tag_types = [re.search("\w+", tag).group() for tag in suffix_tags]  # e.g.['size', 'color']
            regex = "|".join(
                [r"\[{0}=.*?\]".format(tag_type) for tag_type in suffix_tag_types] +
                [r"\<{0}=.*?\>".format(tag_type) for tag_type in suffix_tag_types])  # e.g.'\[size=.*?\]|\[color=.*?\]'
            prefix_tags = list(set(re.findall(regex, text) if regex else []))  # e.g. ['[color=#F7C358]', '[size=40]']
            tags_all = suffix_tags + prefix_tags  # e.g. ['[/color]', '[/size]', '[color=#F7C358]', '[size=40]']
            for tag in tags_all:
                text = text.replace(tag, "<code>{0}</code>".format(tag))

            # 术语预处理
            for glossary_source, glossary_target in glossary_dict.items():
                text = text.replace(glossary_source,
                                    "<start>{0}<middle>{1}<end>".format(glossary_source, glossary_target))
            df.at[index, column] = text
df
```

```python
df.to_excel("實時同步翻譯文本_tagged.xlsx", index=False)
```
