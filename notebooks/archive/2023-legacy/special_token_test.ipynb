```python
from transformers import AutoTokenizer

model_name = "pretrained-models/nllb-200-distilled-600M"
tokenizer = AutoTokenizer.from_pretrained(model_name, tgt_lang=None)
tokenizer
```

```python
additional_special_tokens = tokenizer.additional_special_tokens
additional_special_tokens.extend(["<token>", "</token>"])
additional_special_tokens
```

```python
special_tokens_dict = {'additional_special_tokens': additional_special_tokens}
tokenizer.add_special_tokens(special_tokens_dict)
```

```python
additional_special_tokens
```

```python
tokenizer.all_special_ids
```

```python
from transformers import AutoModelForSeq2SeqLM

model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
# 调整embedding层的大小
model.resize_token_embeddings(len(tokenizer))
```

```python
text = "你好,<token>世界</token>"
inputs = tokenizer(text, return_tensors="pt")
inputs
```

```python
import torch

inputs.attention_mask = torch.tensor([[1, 1, 1, 0, 1, 0, 1]])
inputs.attention_mask
```

```python
translated_tokens = model.generate(
    **inputs, forced_bos_token_id=tokenizer.lang_code_to_id["kor_Hang"], max_length=30
)
translated_tokens
```

```python
tokenizer.batch_decode(translated_tokens, skip_special_tokens=False)[0]
```
