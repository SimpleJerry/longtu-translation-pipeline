```python
import pickle

# 反序列化
file_name = "not_fanyi_Lang_korea_韩文"
with open("{0}_code_logs.pkl".format(file_name), "rb") as file:
    code_token_logs = pickle.load(file)
```

```python
code_token_logs
```

```python
import pandas as pd

df = pd.read_excel("translation_result_of_fine-tuned-models_nllb-200-1.3B_zh2ko_1107.xlsx")
df
```

```python
for log in code_token_logs:
    index, column, code, token = log
    df.loc[index, "candidates"] = str(df.loc[index, "candidates"]).replace(token, code)
```

```python
df.to_excel("not_fanyi_Lang_korea_韩文_translated.xlsx")
```
