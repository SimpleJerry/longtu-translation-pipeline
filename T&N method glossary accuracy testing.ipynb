```python
import pandas as pd

file_name = "translation_result_of_fine-tuned-models_nllb-200-distilled-600M_zh2ko_1025.csv"
df = pd.read_csv(file_name)
df
```

```python
import re

for index, row in df.iterrows():
    match = re.search("<start>(.*?)<middle>(.*?)<end>", row["source"])
    glossary_zh_CN = match.group(1)
    glossary_ko = match.group(2)
    if glossary_ko in row["candidates"]:
        df.at[index, "matched"] = 1
    else:
        df.at[index, "matched"] = 0
```

```python
df.to_csv(file_name, index=False)
df
```
