```python
import pandas as pd

# df = pd.read_csv("translation_result.csv")
df = pd.read_excel("../translation_result_of_autodl-tmp_nllb-200-1.3B_zh2ko_1107.xlsx")
df["candidates"] = df["candidates"].fillna(value="<pad>")
df
```

```python
# references = df["references"].to_list()
references = [[str(reference)] for reference in df["references"].to_list()]
candidates = [str(candidate) for candidate in df["candidates"].to_list()]
```

```python
from nltk.translate.bleu_score import corpus_bleu

score = corpus_bleu(references, candidates)
score
```
