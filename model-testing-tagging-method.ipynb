Preprocessing

```python
import pandas as pd

df_glossary = pd.read_csv("glossary_all.csv")
df_glossary = df_glossary[["zh-CN", "ko"]].dropna()
df_glossary["zh-CN_length"] = df_glossary["zh-CN"].str.len()
df_glossary = df_glossary.sort_values("zh-CN_length", ascending=False)
df_glossary = df_glossary.drop('zh-CN_length', axis=1)
df_glossary
```

```python
glossary_dict = {index: {"zh-CN": row["zh-CN"], "ko": row["ko"]} for index, row in df_glossary.iterrows()}
glossary_dict
```

```python
import pandas as pd

df = pd.read_csv("태왕_단가_용어집.csv")
df
```

```python
reference = df["zh-CN"].to_list()[:1000]  # 待翻译的句子
translation = df["ko"].to_list()[:1000]  # 标准的翻译
```

```python
reference_pre = []
for text in reference:
    for glossary_id, glossary in glossary_dict.items():
        text = text.replace(glossary["zh-CN"], "<glossary_id={0}>".format(glossary_id))
    reference_pre.append(text)
reference_pre
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 加载model,tokenizer
custom_model_name = "fine-tuned-models/nllb-200-3.3B/zh2ko_0913"

model = AutoModelForSeq2SeqLM.from_pretrained(custom_model_name)
tokenizer = AutoTokenizer.from_pretrained(custom_model_name)
```

```python
from transformers import pipeline

translator = pipeline("translation", model=model, tokenizer=tokenizer,
                      src_lang=tokenizer.src_lang,
                      tgt_lang=tokenizer.tgt_lang,
                      device_map="auto", max_length=400)
# translator = pipeline("translation", model=model, tokenizer=tokenizer,
#                       src_lang=tokenizer.src_lang,
#                       tgt_lang=tokenizer.tgt_lang,
#                       device="cuda:0", max_length=400)
```

```python
translated = translator(reference_pre)
translated_text = [item['translation_text'] for item in translated]
translated_text
```

```python
import re

for index in range(len(translated_text)):
    temp = translated_text[index]
    glossary_ids = [int(glossary_id) for glossary_id in re.findall(r"<glossary_id=(\d+)>", temp)]
    for glossary_id in glossary_ids:
        temp = temp.replace("<glossary_id={0}>".format(glossary_id), glossary_dict[glossary_id]["ko"])
    translated_text[index] = temp
translated_text
```

```python
df_compare = pd.DataFrame({
    "reference": reference,
    "translation": translation,
    "translated": translated_text,
})
df_compare
```

```python
df_compare.to_csv("translation_result.csv", index=False)
```
