```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 加载model,tokenizer
custom_model_name = "fine-tuned-models/nllb-200-1.3B/zh2ko_0907"

model = AutoModelForSeq2SeqLM.from_pretrained(custom_model_name)
tokenizer = AutoTokenizer.from_pretrained(custom_model_name)
```

```python
from transformers import pipeline

translator = pipeline("translation", model=model, tokenizer=tokenizer,
                      src_lang=tokenizer.src_lang,
                      tgt_lang=tokenizer.tgt_lang,
                      device="cuda:0", max_length=400)
```

```python
import pandas as pd

df = pd.read_excel("검과마법 글로벌 번역 요청_230907.xlsx")

reference = df["중국어"].to_list()[:10]  # 待翻译的句子
```

```python
from tqdm.notebook import tqdm

# 进行翻译并提取结果
translated = [translator(item)[0] for item in tqdm(reference)]
translated_text = [item['translation_text'] for item in translated]
```

```python
df_compare = pd.DataFrame({
    "源句": reference,
    "实际结果": translated_text,
})
df_compare
```

```python
df_compare.to_excel("translation_result.xlsx", index=False)
```
